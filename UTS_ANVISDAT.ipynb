```python
import pandas as pd
import numpy as np

# 1. Memuat dataset menggunakan path absolut yang benar
df = pd.read_csv('/hotel_bookings.csv')
print(f"Jumlah baris awal: {df.shape[0]}, Jumlah kolom: {df.shape[1]}")

# 2. Identifikasi Missing Value
print("\n=== 1. INFORMASI MISSING VALUE ===")
missing_val = df.isnull().sum()
print(missing_val[missing_val > 0])
```

```python
# 3. Periksa Data Duplikat
duplicate_count = df.duplicated().sum()
print(f"\n=== 2. DATA DUPLIKAT ===")
print(f"Jumlah baris duplikat ditemukan: {duplicate_count}")
```

```python
# 4. Identifikasi Outlier pada Variabel Numerik (Contoh: 'adr' dan 'lead_time')
print("\n=== 3. STATISTIK DESKRIPTIF UNTUK IDENTIFIKASI OUTLIER ===")
print(df[['lead_time', 'adr', 'stays_in_weekend_nights']].describe())
```

```python
# Membuat salinan dataframe untuk proses cleaning
df_clean = df.copy()

# 1. Menangani Missing Value (Imputasi)
df_clean['children'] = df_clean['children'].fillna(0)
df_clean['country'] = df_clean['country'].fillna('Unknown')
df_clean['agent'] = df_clean['agent'].fillna(0)
df_clean['company'] = df_clean['company'].fillna(0)

# 2. Menghapus Data Duplikat
initial_rows = df_clean.shape[0]
df_clean = df_clean.drop_duplicates()
duplicate_removed = initial_rows - df_clean.shape[0]

# 3. Menangani Anomali Outlier Ekstrim (Contoh: Menghapus tarif adr yang bernilai negatif)
df_clean = df_clean[df_clean['adr'] >= 0]

# 4. Verifikasi Hasil Akhir Preprocessing
print("=== HASIL AKHIR PREPROCESSING ===")
print(f"Jumlah baris awal: {initial_rows}")
print(f"Jumlah baris duplikat yang dihapus: {duplicate_removed}")
print(f"Jumlah baris akhir dataset bersih: {df_clean.shape[0]}")
print(f"Total missing value tersisa: {df_clean.isnull().sum().sum()}")
```

```python
import pandas as pd
import numpy as np

# Menentukan variabel numerik yang relevan untuk dianalisis
num_cols = ['lead_time', 'adr', 'stays_in_weekend_nights', 'stays_in_week_nights', 'adults']

# Menghitung mean, std, min, max dari fungsi describe()
desc_num = df_clean[num_cols].describe().T[['mean', 'std', 'min', 'max']].copy()

# Menghitung median (nilai tengah / kuartil 50%)
desc_num['median'] = df_clean[num_cols].median()

# Menghitung modus (nilai yang paling sering muncul)
modes = []
for col in num_cols:
    mode_val = df_clean[col].mode()
    modes.append(mode_val.iloc[0] if not mode_val.empty else np.nan)
desc_num['modus'] = modes

# Menyusun urutan kolom agar sesuai: mean, median, modus, min, max, std
desc_num = desc_num[['mean', 'median', 'modus', 'min', 'max', 'std']]

print("=== TABEL STATISTIK DESKRIPTIF VARIABEL NUMERIK ===")
display(desc_num)
```

```python
import pandas as pd

# Menentukan variabel kategorikal utama yang relevan untuk dianalisis
cat_cols = ['hotel', 'market_segment', 'customer_type', 'deposit_type']

# Melakukan perulangan untuk menampilkan frekuensi dan persentase tiap variabel kategorikal
for col in cat_cols:
    print(f"\n=== DISTRIBUSI FREKUENSI & PERSENTASE: {col.upper()} ===")
    freq = df_clean[col].value_counts()
    perc = df_clean[col].value_counts(normalize=True) * 100
    summary_df = pd.DataFrame({
        'Frekuensi (Jumlah)': freq,
        'Persentase (%)': perc.round(2)
    })
    display(summary_df)
```

```python
import pandas as pd

# === TUGAS 3: ANALISIS DISTRIBUSI DATA (Skewness / Kemiringan) ===
print("=== ANALISIS KEMIRINGAN DISTRIBUSI (SKEWNESS) ===")
num_cols = ['lead_time', 'adr', 'stays_in_weekend_nights', 'stays_in_week_nights', 'adults']
skewness_vals = df_clean[num_cols].skew()
print(skewness_vals)
```

```python
import pandas as pd

# 1. Membandingkan rata-rata berdasarkan Tipe Hotel (City Hotel vs Resort Hotel)
print("=== PERBEDAAN BERDASARKAN TIPE HOTEL ===")
hotel_group = df_clean.groupby('hotel')[['adr', 'lead_time', 'stays_in_week_nights']].mean()
display(hotel_group)

# 2. Membandingkan rata-rata berdasarkan Segmen Pasar (Market Segment)
print("\n=== PERBEDAAN BERDASARKAN SEGMEN PASAR (MARKET SEGMENT) ===")
market_group = df_clean.groupby('market_segment')[['adr', 'lead_time']].mean().sort_values(by='adr', ascending=False)
display(market_group)
```
