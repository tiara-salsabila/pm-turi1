LATIHAN

```python
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}  # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]   # fitur
y = df['harga']                     # label

# Label numerik kontinu -> REGRESI
print(df['harga'].tolist())
# Label kategori -> KLASIFIKASI
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))
```

Supervised yaitu data berlabel yang pastinya memiliki kunci jawaban sebagai ukuran tingkat keakuratan. Unsupervised yaitu data tanpa label untuk menemukan pola tersembunyi dalam menghasilkan jawaban. Reinforcement learning yaitu system diberi reward sebagai penentu benar atau salahnya.

Data latih dan data uji perlu dibagi karena data latih hanya digunakan untuk melatih pola dalam menghasilkan kunci jawaban, sedangkan data uji digunakan sebagai standar apakah mesin sudah menghasilkan jawaban yang akurat atau tidak setelah belajar dari data latih.
